```python
import torch
import torch.nn as nn

x = torch.randn(3, 4) #it make a random input

Q = nn.Linear(4, 4)(x) #it is query
K = nn.Linear(4, 4)(x) #it is key
V = nn.Linear(4, 4)(x) #it is value

scores = Q @ K.T #it check how tokens will relate

scores = scores / (4 ** 0.5) #it keep scores from getting too big

weights = torch.softmax(scores, dim=-1) #it turn scores into weights

output = weights @ V #the final output
```

```python
print(output)
```

Self-attention is a mechanism that helps each word in a sentence understand its relationship with other words. It uses Query, Key and Value to calculate which words are more important and assigns attention weights accordingly. This helps the model understand context rather than processing each word independently.
